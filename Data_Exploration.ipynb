```python
# We are using ffill for filling the nan values because there is multiple rows representing symptoms for same disease

df['Disease'] = df['Disease'].ffill()
```

```python
# We are using ffill for filling the nan values in because there is multiple rows representing symptoms for same Count of Disease Occurrence

df['Count of Disease Occurrence'] = df['Count of Disease Occurrence'].ffill()
```

```python
df[df['Symptom'].isnull()]
```

```python
df = df.dropna(subset = ('Symptom'))
```
