```python
# Importing required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside the notebook
%matplotlib inline

# Improve plot appearance
plt.style.use("ggplot")
```

```python
# Importing cleaned dataset

file_path = '/content/drive/MyDrive/AI_Healthcare_Diagnosis_system/cleaned_data.xlsx'

df = pd.read_excel(file_path)

df.head()
```

```python
print("Dataset Shape:", df.shape)
```

```python
print("Total Diseases:", df["Disease"].nunique())
```

```python
print("Total Unique Symptoms:", df["Symptom"].nunique())
```

```python
# Symptoms per Disease
Symp_count = df.groupby('Disease')['Symptom'].count()

Symp_count
```

```python
# Top 10 Diseases with the Most Symptoms
top10 = Symp_count.sort_values(ascending= False).head(10)

top10
```

```python
# Visualizing top10 disease

plt.figure(figsize=(12,6))
top10.plot(kind="bar")
plt.title("Top 10 Diseases with the Most Symptoms")
plt.xlabel("Disease")
plt.ylabel("Number of Symptoms")
plt.xticks(rotation=90)
plt.show()
```

```python
# Most Common Symptoms

common_symptomps = df['Symptom'].value_counts()
common_symptomps
```

```python
plt.figure(figsize=(12,6))
common_symptomps.head(10).plot(kind="bar", color="orange")
plt.title("Top 10 Most Common Symptoms")
plt.xlabel("Symptoms")
plt.ylabel("Frequency")
plt.xticks(rotation=90)
plt.show()
```
