```python
import pandas as pd
import numpy as np

from sklearn.metrics.pairwise import cosine_similarity
```

cosine_similarity

This is one of the most popular similarity algorithms.

It compares two vectors.


```python
feature_path = "/content/drive/MyDrive/AI_Healthcare_Diagnosis_system/feature_dataset.xlsx"

df = pd.read_excel(feature_path)

df.head()
df.shape
```

```python
disease = df['Disease']

features = df.drop("Disease", axis=1)
```

```python
disease
```

```python
features
```

```python
# Display all sysmpotoms
symptom_list = list(features.columns)

print("Total Symptoms:", len(symptom_list))
```

```python
user_vector = np.zeros(len(symptom_list))
```

```python
# Manually Testing if user Input given
selected_symptoms = [
    "UMLS:C0015967_fever",
    "UMLS:C0010200_cough",
    "UMLS:C0042963_vomiting",
    "UMLS:C0018681_headache",
    "UMLS:C0085593_chill"
]
```

```python
# Converting symptomps into vector format
for symptom in selected_symptoms:

    if symptom in symptom_list:

        index = symptom_list.index(symptom)

        user_vector[index] = 1
```

```python
print(user_vector)
```

```python
# Calculating Similarity

similarity_scores = cosine_similarity(
    [user_vector],
    features
)
# compares one user vector against all 134 disease vectors.
```

```python
print(similarity_scores.shape)
```

```python
# Flatten the Scores
similarity_scores = similarity_scores.flatten()

print(similarity_scores)
```

```python
# FInding top 3 Diseases

top3 = np.argsort(similarity_scores)[-3:][::-1]

print(top3)
```

```python
# Display the prediction
for i in top3:
    print("Disease :", disease.iloc[i])
    print("Similarity :", round(similarity_scores[i] * 100, 2), "%")
    print("-" * 50)
```

```python
print(similarity_scores.shape)
```

```python
# Displaying Disease name more user friendly
def clean_disease_name(name):
    # Split multiple disease names
    name = name.split("^")[0]

    # Remove UMLS code
    if "_" in name:
        name = name.split("_", 1)[1]

    return name.title()
```

```python
diseases_clean = disease.apply(clean_disease_name)
```

```python
for i in top3:
    print(diseases_clean.iloc[i])
```

```python
# For feature
def clean_symptom(symptom):
    if "_" in symptom:
        symptom = symptom.split("_", 1)[1]

    return symptom.replace("_", " ").title()
```

```python
clean_symptoms = [clean_symptom(s) for s in symptom_list]

print(clean_symptoms[:10])
```
