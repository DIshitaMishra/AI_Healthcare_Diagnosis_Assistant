```python
import pandas as pd

file_path = '/content/drive/MyDrive/AI_Healthcare_Diagnosis_system/cleaned_data.xlsx'

df = pd.read_excel(file_path)
df.head(10)
```

```python
# Inspecting one disease
df[df['Disease'] == "UMLS:C0011847_diabetes"]
```

```python

disease_symp = df.groupby('Disease')['Symptom'].apply(list)

disease_symp.head(10)
```

```python
disease_symp = disease_symp.reset_index()

disease_symp.head(10)
```

```python
# Count the Diseases
print("Number of Disease:", len(disease_symp))
```

```python
# Checking one disease symptoms list
disease_symp.iloc[0]
```

```python
print("Shape:", disease_symp.shape)
```

```python
# Now we are going to create the feature matrix for the training, extracting the dataset into 1,0 format for all sysmptoms there will be one column associated

all_symptoms = sorted(df['Symptom'].unique())

print("Total Unique Symptoms:", len(all_symptoms))
```

```python
feature_df = pd.DataFrame(0, index = disease_symp.index, columns= all_symptoms)

feature_df.head()
```

```python
for i, symptoms in enumerate(disease_symp["Symptom"]):
    for symptom in symptoms:
        feature_df.loc[i, symptom] = 1
```

```python
feature_df.insert(0, "Disease", disease_symp["Disease"])
```

```python
feature_df.head()
```

```python
print(feature_df.shape)
```

```python
# Saving the Feature matrix file
feature_path = "/content/drive/MyDrive/AI_Healthcare_Diagnosis_system/feature_dataset.xlsx"
feature_df.to_excel(feature_path, index=False)

print("Feature dataset saved successfully!")
```
